### Objective

This notebook generates a realistic dirty version of the Customer Master dataset.

The original Customer_Master.csv remains unchanged and acts as the Master Dataset.

Python is used only to inject planned business data quality issues.

```python
#CELL 8 - INJECT LEADING & TRAILING SPACES

for row in leading_space_rows:

    current_name = customer_dirty.loc[row, "Customer_Name"]

    space_type = random.choice([
        "leading",
        "trailing",
        "both"
    ])

    if space_type == "leading":
        customer_dirty.loc[row, "Customer_Name"] = " " + current_name

    elif space_type == "trailing":
        customer_dirty.loc[row, "Customer_Name"] = current_name + " "

    else:
        customer_dirty.loc[row, "Customer_Name"] = " " + current_name + " "
```

```python
#CELL 13 - INJECT MIXED CAPITALIZATION INTO CUSTOMER_NAME

def random_mixed_case(text):
    """
    Convert alphabetic characters to random upper/lower case.
    Guarantees that the output differs from the input.
    """

    while True:

        mixed = "".join(
            random.choice([char.upper(), char.lower()])
            if char.isalpha()
            else char
            for char in text
        )

        if mixed != text:
            return mixed
```

```python
#CELL 13 - INJECT MIXED CAPITALIZATION INRO Customer_Name

def random_mixed_case(text):
    """
    Convert alphabetic characters to random upper/lower case.
    Non-alphabetic characters (such as spaces) remain unchanged.
    """

    return "".join(
        random.choice([char.upper(), char.lower()]) if char.isalpha() else char
        for char in text
    )


for row in mixed_case_rows:

    original_name = customer_master.loc[row, "Customer_Name"]

    customer_dirty.loc[row, "Customer_Name"] = random_mixed_case(original_name)
```

```python
#CELL 23 - INJECT STATE ABBREVIATIONS

STATE_ABBREVIATIONS = {
    "Andhra Pradesh": "AP",
    "Assam": "AS",
    "Bihar": "BR",
    "Chandigarh": "CH",
    "Delhi": "DL",
    "Gujarat": "GJ",
    "Jharkhand": "JH",
    "Karnataka": "KA",
    "Kerala": "KL",
    "Maharashtra": "MH",
    "Odisha": "OD",
    "Rajasthan": "RJ",
    "Tamil Nadu": "TN",
    "Telangana": "TS",
    "Uttar Pradesh": "UP",
    "West Bengal": "WB"
}

# ============================================================
# Inject State Abbreviations
# ============================================================

for row in state_abbreviation_rows:

    original_state = customer_master.loc[row, "State"]

    customer_dirty.loc[row, "State"] = STATE_ABBREVIATIONS[original_state]
```

```python
#CELL 27 - APPEND DUPLICCATE RECORDS

# ============================================================
# Append Duplicate Records
# ============================================================

duplicate_records = customer_dirty.loc[
    duplicate_rows
].copy()

customer_dirty = pd.concat(
    [customer_dirty, duplicate_records],
    ignore_index=True
)

# ============================================================
# Shuffle Dataset (Realistic Duplicate Distribution)
# ============================================================

customer_dirty = customer_dirty.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```
