## Notebook 04 — Returns_Dirty_Generation.ipynb

### Objective

This notebook generates a dirty version of the Returns dataset for Microsoft Excel data cleaning practice.

The master dataset (`Returns.csv`) remains unchanged throughout the process.

Workflow:

Returns.csv
    ↓
Working Copy
    ↓
Inject Planned Errors
    ↓
Validate Every Error
    ↓
Update Modification Tracker
    ↓
Export Returns_Dirty.csv

This notebook follows the same implementation methodology used in the previous dirty data generation notebooks to ensure consistency across the FictionalCart project.

---

## Frozen Error Plan

| Error Type | Rows |
|------------|-----:|
| Leading Spaces | 8 |
| Trailing Spaces | 8 |
| Double Spaces | 8 |
| ALL CAPS | 8 |
| Lowercase | 8 |
| Mixed Capitalization | 8 |
| Hidden Tabs | 5 |
| Blank Return_Date | 4 |
| Blank Refund_Amount | 4 |
| Blank Return_Reason | 4 |
| Duplicate Records | 10 |

Text Variations have been intentionally skipped because Return_Reason is a controlled business field.

```python
#CELL 2 - IMPORT REQUIRED LIBRARIES

import pandas as pd
import numpy as np
import random

from IPython.display import display
```

```python
#CELL 3 - FIXED RANDOM SEED

RANDOM_SEED = 42

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

print(f"Random Seed Fixed at: {RANDOM_SEED}")
```

```python
#CELL 4 - LOAD MASTER DATASET

returns_master = pd.read_csv("../01_Master_Datasets/Returns.csv")

returns_dirty = returns_master.copy(deep=True)

print("Master Dataset Loaded Successfully.\n")

print(f"Rows    : {len(returns_master)}")
print(f"Columns : {returns_master.shape[1]}")
```

```python
#CELL 5 - INITIAL DATASET VALIDATION

print("=" * 60)
print("RETURNS DATASET INFORMATION")
print("=" * 60)

display(returns_dirty.head())

print("\nColumn Information\n")
print(returns_dirty.dtypes)

print("\nMissing Values\n")
print(returns_dirty.isna().sum())
```

```python
#CELL 6 - INITIALIZE MODIFICATION TRACKER

modified_rows = set()

print("Modification Tracker Initialized.")

print(f"Modified Rows : {len(modified_rows)}")
```

```python
#CELL 7 - LEADING SPACES IMPLEMENTAION

print("=" * 60)
print("Injecting Leading Spaces")
print("=" * 60)

ERROR_COUNT = 8

# Select only rows that have not been modified
available_rows = list(set(returns_dirty.index) - modified_rows)

selected_rows = random.sample(available_rows, ERROR_COUNT)

# Store original values for validation
leading_spaces_original = returns_dirty.loc[
    selected_rows,
    "Return_Reason"
].copy()

# Inject leading spaces
returns_dirty.loc[
    selected_rows,
    "Return_Reason"
] = (
    "   " +
    returns_dirty.loc[selected_rows, "Return_Reason"].astype(str)
)

print(f"Selected Rows : {len(selected_rows)}")
print("Leading spaces injected successfully.")
```

```python
#CELL 8 - LEADING SPACES VALIDATION

print("=" * 60)
print("LEADING SPACES VALIDATION")
print("=" * 60)

actual_count = (
    returns_dirty.loc[selected_rows, "Return_Reason"]
    .str.startswith("   ")
    .sum()
)

print(f"Expected Rows : {ERROR_COUNT}")
print(f"Actual Rows   : {actual_count}")

assert actual_count == ERROR_COUNT

print("\nFirst Five Before vs After\n")

for i in range(min(5, ERROR_COUNT)):
    row = selected_rows[i]

    print(f"Row Index : {row}")

    print("Before :", repr(leading_spaces_original.loc[row]))
    print("After  :", repr(returns_dirty.loc[row, "Return_Reason"]))
    print("-" * 50)

print("\nSample Modified Rows\n")

display(
    returns_dirty.loc[
        selected_rows[:5]
    ]
)

print("SUCCESS: Leading Spaces Validation Passed")
```

```python
#CELL 9 - TRACKER UPDATE

previous_count = len(modified_rows)

modified_rows.update(selected_rows)

current_count = len(modified_rows)

print("=" * 60)
print("TRACKER VALIDATION")
print("=" * 60)

print(f"Previous Count : {previous_count}")
print(f"Rows Added     : {ERROR_COUNT}")
print(f"Current Count  : {current_count}")

assert current_count == previous_count + ERROR_COUNT

print("\nSUCCESS: Modification Tracker Updated Successfully")
```

```python
#CELL 10 - TRAILING SPACES IMPLEMENTATION

print("=" * 60)
print("Injecting Trailing Spaces")
print("=" * 60)

ERROR_COUNT = 8

# Select only rows that have not been modified
available_rows = list(set(returns_dirty.index) - modified_rows)

selected_rows = random.sample(available_rows, ERROR_COUNT)

# Store original values for validation
trailing_spaces_original = returns_dirty.loc[
    selected_rows,
    "Return_Reason"
].copy()

# Inject trailing spaces
returns_dirty.loc[
    selected_rows,
    "Return_Reason"
] = (
    returns_dirty.loc[selected_rows, "Return_Reason"].astype(str)
    + "   "
)

print(f"Selected Rows : {len(selected_rows)}")
print("Trailing spaces injected successfully.")
```

```python
#CELL 11 - TRAILING SPACES VALIDATION

print("=" * 60)
print("TRAILING SPACES VALIDATION")
print("=" * 60)

actual_count = (
    returns_dirty.loc[selected_rows, "Return_Reason"]
    .str.endswith("   ")
    .sum()
)

print(f"Expected Rows : {ERROR_COUNT}")
print(f"Actual Rows   : {actual_count}")

assert actual_count == ERROR_COUNT

print("\nFirst Five Before vs After\n")

for i in range(min(5, ERROR_COUNT)):
    row = selected_rows[i]

    print(f"Row Index : {row}")

    print("Before :", repr(trailing_spaces_original.loc[row]))
    print("After  :", repr(returns_dirty.loc[row, "Return_Reason"]))
    print("-" * 50)

print("\nSample Modified Rows\n")

display(
    returns_dirty.loc[selected_rows[:5]]
)

print("SUCCESS: Trailing Spaces Validation Passed")
```

```python
#CELL 12 - TRACKER UPDATE

previous_count = len(modified_rows)

modified_rows.update(selected_rows)

current_count = len(modified_rows)

print("=" * 60)
print("TRACKER VALIDATION")
print("=" * 60)

print(f"Previous Count : {previous_count}")
print(f"Rows Added     : {ERROR_COUNT}")
print(f"Current Count  : {current_count}")

assert current_count == previous_count + ERROR_COUNT

print("\nSUCCESS: Modification Tracker Updated Successfully")
```

```python
#CELL 13 - DOUBLE SPACES IMPLEMENTATION

print("=" * 60)
print("Injecting Double Spaces")
print("=" * 60)

ERROR_COUNT = 8

# Select only rows that have not been modified
available_rows = list(set(returns_dirty.index) - modified_rows)

selected_rows = random.sample(available_rows, ERROR_COUNT)

# Store original values for validation
double_spaces_original = returns_dirty.loc[
    selected_rows,
    "Return_Reason"
].copy()

# Inject multiple spaces between words
returns_dirty.loc[
    selected_rows,
    "Return_Reason"
] = (
    returns_dirty.loc[selected_rows, "Return_Reason"]
    .astype(str)
    .str.replace(" ", "   ", n=1, regex=False)
)

print(f"Selected Rows : {len(selected_rows)}")
print("Double spaces injected successfully.")
```

```python
#CELL 14 - DOUBLE SPACES VALIDATION

print("=" * 60)
print("DOUBLE SPACES VALIDATION")
print("=" * 60)

actual_count = (
    returns_dirty.loc[selected_rows, "Return_Reason"]
    .str.contains("   ", regex=False)
    .sum()
)

print(f"Expected Rows : {ERROR_COUNT}")
print(f"Actual Rows   : {actual_count}")

assert actual_count == ERROR_COUNT

print("\nFirst Five Before vs After\n")

for i in range(min(5, ERROR_COUNT)):
    row = selected_rows[i]

    print(f"Row Index : {row}")

    print("Before :", repr(double_spaces_original.loc[row]))
    print("After  :", repr(returns_dirty.loc[row, "Return_Reason"]))
    print("-" * 50)

print("\nSample Modified Rows\n")

display(
    returns_dirty.loc[selected_rows[:5]]
)

print("SUCCESS: Double Spaces Validation Passed")
```

```python
#CELL 16 - ALL CAPS

print("=" * 60)
print("Injecting ALL CAPS")
print("=" * 60)

ERROR_COUNT = 8

# Select only rows that have not been modified
available_rows = list(set(returns_dirty.index) - modified_rows)

selected_rows = random.sample(available_rows, ERROR_COUNT)

# Store original values for validation
all_caps_original = returns_dirty.loc[
    selected_rows,
    "Return_Reason"
].copy()

# Convert text to ALL CAPS
returns_dirty.loc[
    selected_rows,
    "Return_Reason"
] = (
    returns_dirty.loc[selected_rows, "Return_Reason"]
    .astype(str)
    .str.upper()
)

print(f"Selected Rows : {len(selected_rows)}")
print("ALL CAPS injected successfully.")
```

```python
#CELL 17 - ALL CAPS

print("=" * 60)
print("ALL CAPS VALIDATION")
print("=" * 60)

actual_count = (
    returns_dirty.loc[selected_rows, "Return_Reason"]
    .eq(
        returns_dirty.loc[selected_rows, "Return_Reason"].str.upper()
    )
    .sum()
)

print(f"Expected Rows : {ERROR_COUNT}")
print(f"Actual Rows   : {actual_count}")

assert actual_count == ERROR_COUNT

print("\nFirst Five Before vs After\n")

for i in range(min(5, ERROR_COUNT)):
    row = selected_rows[i]

    print(f"Row Index : {row}")

    print("Before :", repr(all_caps_original.loc[row]))
    print("After  :", repr(returns_dirty.loc[row, "Return_Reason"]))
    print("-" * 50)

print("\nSample Modified Rows\n")

display(
    returns_dirty.loc[selected_rows[:5]]
)

print("SUCCESS: ALL CAPS Validation Passed")
```

```python
#CELL 22 - MIXED CAPITALIZATION IMPLEMENTATION

print("=" * 60)
print("Injecting Mixed Capitalization")
print("=" * 60)

ERROR_COUNT = 8

# Select only rows that have not been modified
available_rows = list(set(returns_dirty.index) - modified_rows)

selected_rows = random.sample(available_rows, ERROR_COUNT)

# Store original values for validation
mixed_case_original = returns_dirty.loc[
    selected_rows,
    "Return_Reason"
].copy()


def alternating_case(text):
    result = []

    for i, char in enumerate(str(text)):
        if char.isalpha():
            if i % 2 == 0:
                result.append(char.upper())
            else:
                result.append(char.lower())
        else:
            result.append(char)

    return "".join(result)


returns_dirty.loc[
    selected_rows,
    "Return_Reason"
] = (
    returns_dirty.loc[selected_rows, "Return_Reason"]
    .apply(alternating_case)
)

print(f"Selected Rows : {len(selected_rows)}")
print("Mixed capitalization injected successfully.")
```

```python
#CELL 23 - MIXED CAPITALIZATION VALIDATION

print("=" * 60)
print("MIXED CAPITALIZATION VALIDATION")
print("=" * 60)

expected_values = mixed_case_original.apply(alternating_case)

actual_count = (
    returns_dirty.loc[selected_rows, "Return_Reason"]
    .eq(expected_values)
    .sum()
)

print(f"Expected Rows : {ERROR_COUNT}")
print(f"Actual Rows   : {actual_count}")

assert actual_count == ERROR_COUNT

print("\nFirst Five Before vs After\n")

for i in range(min(5, ERROR_COUNT)):
    row = selected_rows[i]

    print(f"Row Index : {row}")

    print("Before :", repr(mixed_case_original.loc[row]))
    print("After  :", repr(returns_dirty.loc[row, "Return_Reason"]))
    print("-" * 50)

print("\nSample Modified Rows\n")

display(
    returns_dirty.loc[selected_rows[:5]]
)

print("SUCCESS: Mixed Capitalization Validation Passed")
```

```python
#CELL 37 - DUPLICATE RECORDS IMPLEMENTATION

print("=" * 60)
print("Injecting Duplicate Records")
print("=" * 60)

DUPLICATE_COUNT = 10

# Select rows to duplicate
duplicate_rows = random.sample(list(returns_dirty.index), DUPLICATE_COUNT)

# Store originals for validation
duplicate_records = returns_dirty.loc[
    duplicate_rows
].copy()

# Append duplicate rows
returns_dirty = pd.concat(
    [returns_dirty, duplicate_records],
    ignore_index=True
)

# Shuffle entire dataset
returns_dirty = (
    returns_dirty
    .sample(frac=1, random_state=RANDOM_SEED)
    .reset_index(drop=True)
)

print(f"Duplicate Rows Selected : {DUPLICATE_COUNT}")
print("Duplicates appended successfully.")
print("Dataset shuffled successfully.")
```
