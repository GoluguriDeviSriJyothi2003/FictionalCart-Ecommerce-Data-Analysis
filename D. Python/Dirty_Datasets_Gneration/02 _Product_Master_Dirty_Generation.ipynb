```python
#CELL IMPORT REQUIRED LIBRARIES

import pandas as pd
import numpy as np
import random
import warnings
```

```python
#CELL 2 - CONFIGURATION NOTEBOOK ENVIRONMENT

# Fixed random seed for reproducibility
SEED = 42

random.seed(SEED)
np.random.seed(SEED)

# Suppress unnecessary warnings
warnings.filterwarnings("ignore")

# Pandas display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 150)
pd.set_option("display.max_colwidth", None)

print("=" * 60)
print("Notebook Configuration Completed")
print(f"Random Seed : {SEED}")
print("=" * 60)
```

```python
#CELL 3 - READ THE MASTER DATASET AND CREATE AN IN-MEMORY COPY

product_master = pd.read_csv("../01_Master_Datasets/Product_Master.csv")

# Create an independent copy for dirty data generation
product_dirty = product_master.copy()

print("=" * 60)
print("Product Master Dataset Loaded Successfully")
print("=" * 60)
print(f"Rows    : {product_master.shape[0]}")
print(f"Columns : {product_master.shape[1]}")
```

```python
#CELL 4 - VALIDATE THE MASTER DATASET AND WORKING COPY

print("=" * 60)
print("DATASET VALIDATION")
print("=" * 60)

print(f"Master Shape : {product_master.shape}")
print(f"Dirty Shape  : {product_dirty.shape}")

print("\nDatasets Identical :", product_master.equals(product_dirty))
```

```python
#CELL 5 - REVIEW PRODUCT DATASET SCHEMA
# ============================================================
# Review Product Dataset Schema
# ============================================================

print("=" * 60)
print("PRODUCT MASTER COLUMNS")
print("=" * 60)

for i, column in enumerate(product_dirty.columns, start=1):
    print(f"{i}. {column}")
```

```python
#CELL 6 - INITIALIZE MODIFICATION TRACKER

modified_rows = set()

print("=" * 60)
print("MODIFICATION TRACKER INITIALIZED")
print("=" * 60)
print(f"Modified Rows : {len(modified_rows)}")
```

```python
#CELL 7 - SELECT ROWS FOR LEADING & TRAILING SPACES

num_space_errors = 20

space_rows = random.sample(product_dirty.index.tolist(), num_space_errors)

print("=" * 60)
print("LEADING & TRAILING SPACES")
print("=" * 60)
print(f"Rows Selected : {len(space_rows)}")
print("\nSelected Row Indices:")
print(sorted(space_rows))
```

```python
#CELL 8 - INJECT LEADING & TRAILING SPACES

# Store original values for validation
space_original = product_dirty.loc[space_rows, "Product_Name"].copy()

# Inject spaces
for row in space_rows:

    value = str(product_dirty.at[row, "Product_Name"])

    if random.random() < 0.5:
        product_dirty.at[row, "Product_Name"] = " " + value
    else:
        product_dirty.at[row, "Product_Name"] = value + " "

print("=" * 60)
print("Leading & Trailing Spaces Injected Successfully")
print("=" * 60)
print(f"Modified Records : {len(space_rows)}")
```

```python
#CELL 9 - VALIDATION

print("=" * 60)
print("VALIDATION - LEADING & TRAILING SPACES")
print("=" * 60)

for row in space_rows[:5]:

    print(f"\nRow Index : {row}")

    print("Before :", repr(space_original.loc[row]))
    print("After  :", repr(product_dirty.at[row, "Product_Name"]))
```

```python
#CELL 10 - UPDATE MODIFICATION TRACKER

modified_rows.update(space_rows)

print("=" * 60)
print("MODIFICATION TRACKER UPDATED")
print("=" * 60)
print(f"Total Modified Rows : {len(modified_rows)}")
```

```python
#CELL 11 - SELECT ROWS FOR DOUBLE SPACES

num_double_space_errors = 15

# Select only previously unmodified rows
available_rows = list(set(product_dirty.index) - modified_rows)

double_space_rows = random.sample(
    available_rows,
    num_double_space_errors
)

print("=" * 60)
print("DOUBLE SPACES")
print("=" * 60)
print(f"Rows Selected : {len(double_space_rows)}")
print("\nSelected Row Indices:")
print(sorted(double_space_rows))
```

```python
#CELL 12 - INJECT DOUBLE SPACES

# Store original values for validation
double_space_original = product_dirty.loc[
    double_space_rows,
    "Product_Name"
].copy()

# Inject double spaces
for row in double_space_rows:

    value = str(product_dirty.at[row, "Product_Name"])

    # Replace only the first single space with a double space
    if " " in value:
        product_dirty.at[row, "Product_Name"] = value.replace(" ", "  ", 1)

print("=" * 60)
print("Double Spaces Injected Successfully")
print("=" * 60)
print(f"Modified Records : {len(double_space_rows)}")
```

```python
#CELL 13 - VALIDATION

print("=" * 60)
print("VALIDATION - DOUBLE SPACES")
print("=" * 60)

for row in double_space_rows[:5]:

    print(f"\nRow Index : {row}")

    print("Before :", repr(double_space_original.loc[row]))
    print("After  :", repr(product_dirty.at[row, "Product_Name"]))
```

```python
#CELL 14 - UPDATE MODIFICATION TRACKER

modified_rows.update(double_space_rows)

print("=" * 60)
print("MODIFICATION TRACKER UPDATED")
print("=" * 60)
print(f"Total Modified Rows : {len(modified_rows)}")
```

```python
#CELL 15 - SELECT ROWS FOR ALL CAPS

num_uppercase_errors = 12

# Select only previously unmodified rows
available_rows = list(set(product_dirty.index) - modified_rows)

uppercase_rows = random.sample(
    available_rows,
    num_uppercase_errors
)

print("=" * 60)
print("ALL CAPS")
print("=" * 60)
print(f"Rows Selected : {len(uppercase_rows)}")
print("\nSelected Row Indices:")
print(sorted(uppercase_rows))
```

```python
#CELL 16 - INJECT ALL CAPS

# Store original values for validation
uppercase_original = product_dirty.loc[
    uppercase_rows,
    "Product_Name"
].copy()

# Convert Product_Name to ALL CAPS
for row in uppercase_rows:

    value = str(product_dirty.at[row, "Product_Name"])

    product_dirty.at[row, "Product_Name"] = value.upper()

print("=" * 60)
print("ALL CAPS Injected Successfully")
print("=" * 60)
print(f"Modified Records : {len(uppercase_rows)}")
```

```python
#CELL 17 - VALIDATION

print("=" * 60)
print("VALIDATION - ALL CAPS")
print("=" * 60)

for row in uppercase_rows[:5]:

    print(f"\nRow Index : {row}")

    print("Before :", repr(uppercase_original.loc[row]))
    print("After  :", repr(product_dirty.at[row, "Product_Name"]))
```

```python
#CELL 18 - UPDATE MODIFICATIOON TRACKER

modified_rows.update(uppercase_rows)

print("=" * 60)
print("MODIFICATION TRACKER UPDATED")
print("=" * 60)
print(f"Total Modified Rows : {len(modified_rows)}")
```

```python
#CELL 19 - SELECT ROWS FOR ALL LOWERCASE

num_lowercase_errors = 12

# Select only previously unmodified rows
available_rows = list(set(product_dirty.index) - modified_rows)

lowercase_rows = random.sample(
    available_rows,
    num_lowercase_errors
)

print("=" * 60)
print("ALL LOWERCASE")
print("=" * 60)
print(f"Rows Selected : {len(lowercase_rows)}")
print("\nSelected Row Indices:")
print(sorted(lowercase_rows))
```

```python
#CELL 20 - INJECT ALL LOWERCASE

# Store original values for validation
lowercase_original = product_dirty.loc[
    lowercase_rows,
    "Product_Name"
].copy()

# Convert Product_Name to lowercase
for row in lowercase_rows:

    value = str(product_dirty.at[row, "Product_Name"])

    product_dirty.at[row, "Product_Name"] = value.lower()

print("=" * 60)
print("all lowercase Injected Successfully")
print("=" * 60)
print(f"Modified Records : {len(lowercase_rows)}")
```

```python
#CELL 21 - VALIDATION

print("=" * 60)
print("VALIDATION - all lowercase")
print("=" * 60)

for row in lowercase_rows[:5]:

    print(f"\nRow Index : {row}")

    print("Before :", repr(lowercase_original.loc[row]))
    print("After  :", repr(product_dirty.at[row, "Product_Name"]))
```

```python
#CELL 22 - UPDATE MODIFICATION TRACKER

modified_rows.update(lowercase_rows)

print("=" * 60)
print("MODIFICATION TRACKER UPDATED")
print("=" * 60)
print(f"Total Modified Rows : {len(modified_rows)}")
```

```python
#CELL 23 - SELECT ROWS FOR MIXED CAPITALIZATION

num_mixed_case_errors = 10

available_rows = list(set(product_dirty.index) - modified_rows)

mixed_case_rows = random.sample(
    available_rows,
    num_mixed_case_errors
)

print("=" * 60)
print("MIXED CAPITALIZATION")
print("=" * 60)
print(f"Rows Selected : {len(mixed_case_rows)}")
print("\nSelected Row Indices:")
print(sorted(mixed_case_rows))
```

```python
#CELL 24 - INJECT MIXED CAPITALIZATION

# Store original values for validation
mixed_case_original = product_dirty.loc[
    mixed_case_rows,
    "Product_Name"
].copy()


# Function to create realistic mixed capitalization
def random_case(text):

    new_text = ""

    for ch in str(text):

        if ch.isalpha():

            if random.random() < 0.35:
                new_text += ch.upper()
            else:
                new_text += ch.lower()

        else:
            new_text += ch

    return new_text


# Inject mixed capitalization
for row in mixed_case_rows:

    product_dirty.at[row, "Product_Name"] = random_case(
        product_dirty.at[row, "Product_Name"]
    )


print("=" * 60)
print("Mixed Capitalization Injected Successfully")
print("=" * 60)
print(f"Modified Records : {len(mixed_case_rows)}")
```

```python
#CELL 25 - VALIDATION

print("=" * 60)
print("VALIDATION - MIXED CAPITALIZATION")
print("=" * 60)

for row in mixed_case_rows[:5]:

    print(f"\nRow Index : {row}")

    print("Before :", repr(mixed_case_original.loc[row]))
    print("After  :", repr(product_dirty.at[row, "Product_Name"]))
```

```python
#CELL 26 - UPDATE TRACKER

modified_rows.update(mixed_case_rows)

print("=" * 60)
print("MODIFICATION TRACKER UPDATED")
print("=" * 60)
print(f"Total Modified Rows : {len(modified_rows)}")
```

```python
#CELL 27 - SELECT ROWS FOR HIDDEN CHARACTERS

num_hidden_char_errors = 12

# Select only previously unmodified rows
available_rows = list(set(product_dirty.index) - modified_rows)

hidden_char_rows = random.sample(
    available_rows,
    num_hidden_char_errors
)

print("=" * 60)
print("HIDDEN CHARACTERS")
print("=" * 60)
print(f"Rows Selected : {len(hidden_char_rows)}")
print("\nSelected Row Indices:")
print(sorted(hidden_char_rows))
```

```python
#CELL 28 - INJECT HIDDEN CHARACTERS

# Store original values for validation
hidden_char_original = product_dirty.loc[
    hidden_char_rows,
    "Product_Name"
].copy()

# Common hidden characters found in exported data
hidden_chars = [
    "\t"
]

# Inject hidden characters
for row in hidden_char_rows:

    value = str(product_dirty.at[row, "Product_Name"])

    hidden_char = random.choice(hidden_chars)

    product_dirty.at[row, "Product_Name"] = value + hidden_char

print("=" * 60)
print("Hidden Characters Injected Successfully")
print("=" * 60)
print(f"Modified Records : {len(hidden_char_rows)}")
```

```python
#CELL 29 - VALIDATION

print("=" * 60)
print("VALIDATION - HIDDEN CHARACTERS")
print("=" * 60)

for row in hidden_char_rows[:5]:

    print(f"\nRow Index : {row}")

    print("Before :", repr(hidden_char_original.loc[row]))
    print("After  :", repr(product_dirty.at[row, "Product_Name"]))
```

```python
#CELL 30 - UPDATE MODIFICATION TRACKER

modified_rows.update(hidden_char_rows)

print("=" * 60)
print("MODIFICATION TRACKER UPDATED")
print("=" * 60)
print(f"Total Modified Rows : {len(modified_rows)}")
```

```python
#CELL 31 - SELECT ROWS FOR BLANK VALUES

num_category_blank = 3
num_price_blank = 2
num_rating_blank = 2
num_date_blank = 2

# Available rows (avoid previously modified rows)
available_rows = list(set(product_dirty.index) - modified_rows)

# Select unique rows for each blank type
category_blank_rows = random.sample(
    available_rows,
    num_category_blank
)

available_rows = list(set(available_rows) - set(category_blank_rows))

price_blank_rows = random.sample(
    available_rows,
    num_price_blank
)

available_rows = list(set(available_rows) - set(price_blank_rows))

rating_blank_rows = random.sample(
    available_rows,
    num_rating_blank
)

available_rows = list(set(available_rows) - set(rating_blank_rows))

date_blank_rows = random.sample(
    available_rows,
    num_date_blank
)

print("="*60)
print("BLANK VALUE ROWS SELECTED")
print("="*60)

print("Category       :", sorted(category_blank_rows))
print("Selling Price  :", sorted(price_blank_rows))
print("Product Rating :", sorted(rating_blank_rows))
print("Launch Date    :", sorted(date_blank_rows))
```

```python
#CELL 32 - INJECT BLANK VALUES

# Store original values
category_blank_original = product_dirty.loc[
    category_blank_rows,
    "Category"
].copy()

price_blank_original = product_dirty.loc[
    price_blank_rows,
    "Selling_Price"
].copy()

rating_blank_original = product_dirty.loc[
    rating_blank_rows,
    "Product_Rating"
].copy()

date_blank_original = product_dirty.loc[
    date_blank_rows,
    "Launch_Date"
].copy()

# Inject blanks
product_dirty.loc[category_blank_rows, "Category"] = np.nan

product_dirty.loc[price_blank_rows, "Selling_Price"] = np.nan

product_dirty.loc[rating_blank_rows, "Product_Rating"] = np.nan

product_dirty.loc[date_blank_rows, "Launch_Date"] = np.nan

print("="*60)
print("Blank Values Injected Successfully")
print("="*60)
print("Total Blank Cells :", 9)
```

```python
#CELL 33 - VALIDATION

print("="*60)
print("VALIDATION - BLANK VALUES")
print("="*60)

print("\nCATEGORY")
for row in category_blank_rows:
    print(
        row,
        repr(category_blank_original.loc[row]),
        "→",
        repr(product_dirty.at[row, "Category"])
    )

print("\nSELLING PRICE")
for row in price_blank_rows:
    print(
        row,
        repr(price_blank_original.loc[row]),
        "→",
        repr(product_dirty.at[row, "Selling_Price"])
    )

print("\nPRODUCT RATING")
for row in rating_blank_rows:
    print(
        row,
        repr(rating_blank_original.loc[row]),
        "→",
        repr(product_dirty.at[row, "Product_Rating"])
    )

print("\nLAUNCH DATE")
for row in date_blank_rows:
    print(
        row,
        repr(date_blank_original.loc[row]),
        "→",
        repr(product_dirty.at[row, "Launch_Date"])
    )
```

```python
#CELL 34 - UPDATE MODIFICATION TRACKER

modified_rows.update(category_blank_rows)
modified_rows.update(price_blank_rows)
modified_rows.update(rating_blank_rows)
modified_rows.update(date_blank_rows)

print("="*60)
print("MODIFICATION TRACKER UPDATED")
print("="*60)
print(f"Total Modified Rows : {len(modified_rows)}")
```

```python
#CELL 35 - SELECT ROWS FOR CATEGORY VARIATIONS

num_category_variations = 12

available_rows = list(set(product_dirty.index) - modified_rows)

category_variation_rows = random.sample(
    available_rows,
    num_category_variations
)

print("=" * 60)
print("CATEGORY VARIATIONS")
print("=" * 60)
print(f"Rows Selected : {len(category_variation_rows)}")
print("\nSelected Row Indices:")
print(sorted(category_variation_rows))
```

```python
#CELL 36 - INJECT CATEGORY VARIATIONS

# Store original values
category_variation_original = product_dirty.loc[
    category_variation_rows,
    "Category"
].copy()

variation_patterns = [
    lambda x: x.lower(),
    lambda x: x.upper(),
    lambda x: " " + x,
    lambda x: x + " ",
]

# Inject variations
for i, row in enumerate(category_variation_rows):

    value = str(product_dirty.at[row, "Category"])

    variation = variation_patterns[
        i % len(variation_patterns)
    ]

    product_dirty.at[row, "Category"] = variation(value)

print("=" * 60)
print("Category Variations Injected Successfully")
print("=" * 60)
print(f"Modified Records : {len(category_variation_rows)}")
```

```python
#CELL 37 - VALIDATION

print("=" * 60)
print("VALIDATION - CATEGORY VARIATIONS")
print("=" * 60)

for row in category_variation_rows[:5]:

    print(f"\nRow Index : {row}")

    print("Before :", repr(category_variation_original.loc[row]))
    print("After  :", repr(product_dirty.at[row, "Category"]))
```

```python
#CELL 38 - UPDATE MODIFICATION TRACKER

modified_rows.update(category_variation_rows)

print("=" * 60)
print("MODIFICATION TRACKER UPDATED")
print("=" * 60)
print(f"Total Modified Rows : {len(modified_rows)}")
```

```python
#CELL 39 - SELECT ROWS FOR DUPLICATE RECORDS

num_duplicate_records = 8

available_rows = list(set(product_dirty.index) - modified_rows)

duplicate_rows = random.sample(
    available_rows,
    num_duplicate_records
)

print("=" * 60)
print("DUPLICATE RECORDS")
print("=" * 60)
print(f"Rows Selected : {len(duplicate_rows)}")
print("\nSelected Row Indices:")
print(sorted(duplicate_rows))
```

```python
#CELL 40 - INJECT DUPLICATE RECORDS

# Store duplicate records
duplicate_records = product_dirty.loc[
    duplicate_rows
].copy()

# Append duplicate rows
product_dirty = pd.concat(
    [product_dirty, duplicate_records],
    ignore_index=True
)

# Shuffle dataset so duplicates are not grouped together
product_dirty = product_dirty.sample(
    frac=1,
    random_state=SEED
).reset_index(drop=True)

print("=" * 60)
print("Duplicate Records Injected Successfully")
print("=" * 60)

print(f"Rows After Duplication : {len(product_dirty)}")
```

```python
#CELL 41 - VALIDATION

print("=" * 60)
print("VALIDATION - DUPLICATE RECORDS")
print("=" * 60)

duplicate_product_ids = product_dirty[
    product_dirty.duplicated(
        subset=["Product_ID"],
        keep=False
    )
].sort_values("Product_ID")

print("Number of Duplicate Product_IDs :",
      duplicate_product_ids["Product_ID"].nunique())

print()

print(
    duplicate_product_ids[
        ["Product_ID", "Product_Name"]
    ]
)
```

```python
#CELL 42 - UPDATE MODIFICATION TRACKER

modified_rows.update(duplicate_rows)

print("=" * 60)
print("MODIFICATION TRACKER UPDATED")
print("=" * 60)
print(f"Total Modified Rows : {len(modified_rows)}")
```

```python
#CELL 43 - FINAL DATASET VALIDATION

print("=" * 60)
print("FINAL DATASET VALIDATION")
print("=" * 60)

print(f"Rows               : {len(product_dirty)}")
print(f"Columns            : {product_dirty.shape[1]}")

duplicate_product_ids = product_dirty.duplicated(
    subset=["Product_ID"]
).sum()

print(f"Duplicate Product_IDs : {duplicate_product_ids}")

print(f"Unique Product_IDs    : {product_dirty['Product_ID'].nunique()}")

print()

if duplicate_product_ids == num_duplicate_records:
    print("✅ Duplicate validation passed")
else:
    print("❌ Duplicate validation failed")

print()

print("Dataset is ready for Microsoft Excel cleaning.")
```

```python
#CELL 44 - DIRTY DATASET SUMMARY

error_summary = {
    "Leading & Trailing Spaces": 20,
    "Double Spaces": 15,
    "ALL CAPS": 12,
    "all lowercase": 12,
    "Mixed Capitalization": 10,
    "Hidden Characters": 12,
    "Blank Values": 9,
    "Category Variations": 12,
    "Duplicate Records": 8
}
```

```python
# ============================================================
# Dirty Dataset Summary
# ============================================================

print("=" * 60)
print("FICTIONALCART PRODUCT DIRTY DATASET SUMMARY")
print("=" * 60)

print(f"Original Rows          : {len(product_master)}")
print(f"Final Rows             : {len(product_dirty)}")
print(f"Columns                : {product_dirty.shape[1]}")
print(f"Unique Modified Rows   : {len(modified_rows)}")

print("-" * 60)

for error, count in error_summary.items():
    print(f"{error:<28}: {count}")

print("-" * 60)

print("Dataset Status         : READY FOR EXCEL CLEANING")

print("=" * 60)
```

```python
#CELL 45 - EXPORT DIRTY DATASET

output_file = "Product_Master_Dirty.csv"

product_dirty.to_csv(
    output_file,
    index=False
)

print("=" * 60)
print("EXPORT COMPLETED")
print("=" * 60)
print(f"File Name : {output_file}")
print(f"Dataset Shape : {product_dirty.shape}")
print("Product_Master_Dirty.csv exported successfully.")
```
