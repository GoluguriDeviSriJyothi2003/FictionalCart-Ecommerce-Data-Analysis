```python
#CELL 1 - IMPORT REQUIRED LIBRARIES

import pandas as pd
import numpy as np
import random

print("=" * 60)
print("FICTIONALCART - ORDERS DIRTY DATA GENERATION")
print("=" * 60)
print("Libraries imported successfully.")
```

```python
#CELL 2 - SET FIXED RANDOM SEED

RANDOM_SEED = 42

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

print("=" * 60)
print("Random seed initialized successfully.")
print(f"Random Seed : {RANDOM_SEED}")
```

```python
#CELL 3 - LOAD ORDERS MASTE DATASET & CREATE WORKING COPY

# File path of master dataset
orders_file = ("../01_Master_Datasets/Orders.csv")

# Read master dataset
orders_master = pd.read_csv(orders_file)

# Create an independent copy for dirty data generation
orders_dirty = orders_master.copy()

# Initialize modified row tracker
modified_rows = set()

print("=" * 60)
print("ORDERS MASTER DATASET LOADED")
print("=" * 60)

print(f"Master Dataset Shape : {orders_master.shape}")
print(f"Working Dataset Shape: {orders_dirty.shape}")

print("\nColumn Names:")
print(list(orders_master.columns))

print("\nModified Rows Tracker Initialized")
print(f"Modified Rows Count : {len(modified_rows)}")
```

```python
#CELL 4 - MASTER DATASET INTEGRITY CHECK & SNAPSHOT

# Store master dataset snapshot
original_shape = orders_master.shape
original_columns = orders_master.columns.tolist()
original_row_count = len(orders_master)
original_order_status = sorted(orders_master["Order_Status"].dropna().unique())

# Validate primary key uniqueness
duplicate_order_ids = orders_master["Order_ID"].duplicated().sum()

print("=" * 60)
print("MASTER DATASET INTEGRITY CHECK")
print("=" * 60)

print(f"Original Shape           : {original_shape}")
print(f"Original Row Count       : {original_row_count}")
print(f"Original Column Count    : {len(original_columns)}")
print(f"Duplicate Order_ID Count : {duplicate_order_ids}")

print("\nOrder_Status Categories:")
for status in original_order_status:
    print(f"• {status}")

# Validation
assert duplicate_order_ids == 0, "Duplicate Order_ID values found in master dataset."

print("\n✅ Master dataset integrity validation passed.")
print("✅ Snapshot successfully stored.")
```

```python
#CELL 5 - LEADING SPACES (IMPLEMENTATION)

# Number of rows to modify
leading_space_count = 40

# Select rows that have not been modified yet
available_rows = list(set(orders_dirty.index) - modified_rows)

selected_rows_leading = random.sample(
    available_rows,
    leading_space_count
)

# Store original values for validation
leading_before = orders_dirty.loc[
    selected_rows_leading,
    "Order_Status"
].copy()

# Inject leading spaces
orders_dirty.loc[
    selected_rows_leading,
    "Order_Status"
] = (
    " " +
    orders_dirty.loc[
        selected_rows_leading,
        "Order_Status"
    ]
)
```

```python
#CELL 6 - LEADING SPACES VALIDATION

print("=" * 60)
print("LEADING SPACES VALIDATION")
print("=" * 60)

print(f"Expected Modified Rows : {leading_space_count}")
print(f"Actual Modified Rows   : {len(selected_rows_leading)}")

# Count rows with leading spaces
leading_space_validation = orders_dirty.loc[
    selected_rows_leading,
    "Order_Status"
].str.startswith(" ").sum()

print(f"Rows Having Leading Spaces : {leading_space_validation}")

print("\nSample Validation (First 5 Rows)")
print("-" * 60)

for row in selected_rows_leading[:5]:
    print(f"Row Index : {row}")
    print(f"Before : {repr(leading_before.loc[row])}")
    print(f"After  : {repr(orders_dirty.loc[row, 'Order_Status'])}")
    print()

# Final Validation
assert len(selected_rows_leading) == leading_space_count, \
    "Incorrect number of rows selected."

assert leading_space_validation == leading_space_count, \
    "Leading space validation failed."

print("✅ Leading spaces successfully injected.")
```

```python
#CELL 7 - LEADING SPACES TRACKER UPDATE

# Update modified rows tracker
modified_rows.update(selected_rows_leading)

print("=" * 60)
print("MODIFIED ROWS TRACKER UPDATED")
print("=" * 60)

print(f"Leading Spaces Added To : {len(selected_rows_leading)} rows")
print(f"Total Modified Rows     : {len(modified_rows)}")

print("\nFirst 10 Modified Row Indexes:")
print(sorted(list(modified_rows))[:10])

# Validation
assert len(modified_rows) == leading_space_count, \
    "Tracker update failed."

print("\n✅ modified_rows tracker updated successfully.")
```

```python
#CELL 8 - TRAILING SPACES IMPLEMENTATION

# Number of rows to modify
trailing_space_count = 40

# Select rows that have not been modified yet
available_rows = list(set(orders_dirty.index) - modified_rows)

selected_rows_trailing = random.sample(
    available_rows,
    trailing_space_count
)

# Store original values for validation
trailing_before = orders_dirty.loc[
    selected_rows_trailing,
    "Order_Status"
].copy()

# Inject trailing spaces
orders_dirty.loc[
    selected_rows_trailing,
    "Order_Status"
] = (
    orders_dirty.loc[
        selected_rows_trailing,
        "Order_Status"
    ] + " "
)
```

```python
#CELL 9 - TRAILING SPACES VALIDATION

print("=" * 60)
print("TRAILING SPACES VALIDATION")
print("=" * 60)

print(f"Expected Modified Rows : {trailing_space_count}")
print(f"Actual Modified Rows   : {len(selected_rows_trailing)}")

# Count rows with trailing spaces
trailing_space_validation = orders_dirty.loc[
    selected_rows_trailing,
    "Order_Status"
].str.endswith(" ").sum()

print(f"Rows Having Trailing Spaces : {trailing_space_validation}")

print("\nSample Validation (First 5 Rows)")
print("-" * 60)

for row in selected_rows_trailing[:5]:
    print(f"Row Index : {row}")
    print(f"Before : {repr(trailing_before.loc[row])}")
    print(f"After  : {repr(orders_dirty.loc[row, 'Order_Status'])}")
    print()

assert len(selected_rows_trailing) == trailing_space_count, \
    "Incorrect number of rows selected."

assert trailing_space_validation == trailing_space_count, \
    "Trailing space validation failed."

print("✅ Trailing spaces successfully injected.")
```

```python
#CELL 10 - TRAILING SPACES TRACKER UPDATE

# Update modified rows tracker
modified_rows.update(selected_rows_trailing)

print("=" * 60)
print("MODIFIED ROWS TRACKER UPDATED")
print("=" * 60)

print(f"Trailing Spaces Added To : {len(selected_rows_trailing)} rows")
print(f"Total Modified Rows      : {len(modified_rows)}")

print("\nFirst 10 Modified Row Indexes:")
print(sorted(list(modified_rows))[:10])

# Validation
expected_modified_rows = leading_space_count + trailing_space_count

assert len(modified_rows) == expected_modified_rows, \
    "Tracker update failed."

print("\n✅ modified_rows tracker updated successfully.")
```

```python
#CELL 11 - ALL CAPS IMPLEMENTATION

# Number of rows to modify
all_caps_count = 30

# Select rows that have not been modified yet
available_rows = list(set(orders_dirty.index) - modified_rows)

selected_rows_all_caps = random.sample(
    available_rows,
    all_caps_count
)

# Store original values for validation
all_caps_before = orders_dirty.loc[
    selected_rows_all_caps,
    "Order_Status"
].copy()

# Convert to ALL CAPS
orders_dirty.loc[
    selected_rows_all_caps,
    "Order_Status"
] = orders_dirty.loc[
    selected_rows_all_caps,
    "Order_Status"
].str.upper()
```

```python
#CELL 12 - ALL CAPS VALIDATION

print("=" * 60)
print("ALL CAPS VALIDATION")
print("=" * 60)

print(f"Expected Modified Rows : {all_caps_count}")
print(f"Actual Modified Rows   : {len(selected_rows_all_caps)}")

# Count rows converted to ALL CAPS
all_caps_validation = (
    orders_dirty.loc[
        selected_rows_all_caps,
        "Order_Status"
    ].str.isupper().sum()
)

print(f"Rows Converted To ALL CAPS : {all_caps_validation}")

print("\nSample Validation (First 5 Rows)")
print("-" * 60)

for row in selected_rows_all_caps[:5]:
    print(f"Row Index : {row}")
    print(f"Before : {repr(all_caps_before.loc[row])}")
    print(f"After  : {repr(orders_dirty.loc[row, 'Order_Status'])}")
    print()

# Validation
assert len(selected_rows_all_caps) == all_caps_count, \
    "Incorrect number of rows selected."

assert all_caps_validation == all_caps_count, \
    "ALL CAPS validation failed."

print("✅ ALL CAPS successfully injected.")
```

```python
#CELL 13 - ALL CAPS TRACKER UPDATE

# Update modified rows tracker
modified_rows.update(selected_rows_all_caps)

print("=" * 60)
print("MODIFIED ROWS TRACKER UPDATED")
print("=" * 60)

print(f"ALL CAPS Added To      : {len(selected_rows_all_caps)} rows")
print(f"Total Modified Rows    : {len(modified_rows)}")

print("\nFirst 10 Modified Row Indexes:")
print(sorted(list(modified_rows))[:10])

# Validation
expected_modified_rows = (
    leading_space_count +
    trailing_space_count +
    all_caps_count
)

assert len(modified_rows) == expected_modified_rows, \
    "Tracker update failed."

print("\n✅ modified_rows tracker updated successfully.")
```

```python
#CELL 14 - ALL LOWERCASE IMPLEMENTATION

# Number of rows to modify
lowercase_count = 30

# Select rows that have not been modified yet
available_rows = list(set(orders_dirty.index) - modified_rows)

selected_rows_lowercase = random.sample(
    available_rows,
    lowercase_count
)

# Store original values for validation
lowercase_before = orders_dirty.loc[
    selected_rows_lowercase,
    "Order_Status"
].copy()

# Convert to lowercase
orders_dirty.loc[
    selected_rows_lowercase,
    "Order_Status"
] = orders_dirty.loc[
    selected_rows_lowercase,
    "Order_Status"
].str.lower()
```

```python
#CELL 15 -ALL LOWERCASE VALIDATION

print("=" * 60)
print("LOWERCASE VALIDATION")
print("=" * 60)

print(f"Expected Modified Rows : {lowercase_count}")
print(f"Actual Modified Rows   : {len(selected_rows_lowercase)}")

# Count lowercase values
lowercase_validation = (
    orders_dirty.loc[
        selected_rows_lowercase,
        "Order_Status"
    ].str.islower().sum()
)

print(f"Rows Converted To Lowercase : {lowercase_validation}")

print("\nSample Validation (First 5 Rows)")
print("-" * 60)

for row in selected_rows_lowercase[:5]:
    print(f"\nRow Index : {row}")
    print(f"Before : {repr(lowercase_before.loc[row])}")
    print(f"After  : {repr(orders_dirty.loc[row, 'Order_Status'])}")

    print("\nComplete Row After Modification")
    print(orders_dirty.loc[[row]])

# Validation
assert len(selected_rows_lowercase) == lowercase_count, \
    "Incorrect number of rows selected."

assert lowercase_validation == lowercase_count, \
    "Lowercase validation failed."

print("\n✅ Lowercase values successfully injected.")
```

```python
#CELL 16 - ALL LOWERCASE TRACKER UPDATE

# Update modified rows tracker
modified_rows.update(selected_rows_lowercase)

print("=" * 60)
print("MODIFIED ROWS TRACKER UPDATED")
print("=" * 60)

print(f"Lowercase Added To     : {len(selected_rows_lowercase)} rows")
print(f"Total Modified Rows    : {len(modified_rows)}")

print("\nFirst 10 Modified Row Indexes:")
print(sorted(list(modified_rows))[:10])

# Validation
expected_modified_rows = (
    leading_space_count +
    trailing_space_count +
    all_caps_count +
    lowercase_count
)

assert len(modified_rows) == expected_modified_rows, \
    "Tracker update failed."

print("\n✅ modified_rows tracker updated successfully.")
```

```python
#CELL 17 - MIXED CAPITALIZATION IMPLEMENTATION

# Number of rows to modify
mixed_case_count = 30

# Select rows that have not been modified yet
available_rows = list(set(orders_dirty.index) - modified_rows)

selected_rows_mixed_case = random.sample(
    available_rows,
    mixed_case_count
)

# Store original values for validation
mixed_case_before = orders_dirty.loc[
    selected_rows_mixed_case,
    "Order_Status"
].copy()


# Function to create realistic mixed capitalization
def create_mixed_case(text):
    result = ""

    for i, char in enumerate(text):
        if i % 2 == 0:
            result += char.lower()
        else:
            result += char.upper()

    return result


# Inject mixed capitalization
orders_dirty.loc[
    selected_rows_mixed_case,
    "Order_Status"
] = orders_dirty.loc[
    selected_rows_mixed_case,
    "Order_Status"
].apply(create_mixed_case)
```

```python
#CELL 18 - MIXED CAPITALIZATION VALIDATION

print("=" * 60)
print("MIXED CAPITALIZATION VALIDATION")
print("=" * 60)

print(f"Expected Modified Rows : {mixed_case_count}")
print(f"Actual Modified Rows   : {len(selected_rows_mixed_case)}")

# Validation count
mixed_case_validation = (
    orders_dirty.loc[
        selected_rows_mixed_case,
        "Order_Status"
    ] != mixed_case_before
).sum()

print(f"Rows Converted To Mixed Capitalization : {mixed_case_validation}")

print("\nSample Validation (First 5 Rows)")
print("-" * 60)

for row in selected_rows_mixed_case[:5]:

    print(f"\nRow Index : {row}")
    print(f"Before : {repr(mixed_case_before.loc[row])}")
    print(f"After  : {repr(orders_dirty.loc[row, 'Order_Status'])}")

    print("\nComplete Row After Modification")
    print(orders_dirty.loc[[row]])

assert len(selected_rows_mixed_case) == mixed_case_count, \
    "Incorrect number of rows selected."

assert mixed_case_validation == mixed_case_count, \
    "Mixed capitalization validation failed."

print("\n✅ Mixed capitalization successfully injected.")
```

```python
#CELL 19 - MIXED CAPITALIZATION TRACKER UPDATE

# Update modified rows tracker
modified_rows.update(selected_rows_mixed_case)

print("=" * 60)
print("MODIFIED ROWS TRACKER UPDATED")
print("=" * 60)

print(f"Mixed Capitalization Added To : {len(selected_rows_mixed_case)} rows")
print(f"Total Modified Rows          : {len(modified_rows)}")

print("\nFirst 10 Modified Row Indexes:")
print(sorted(list(modified_rows))[:10])

# Validation
expected_modified_rows = (
    leading_space_count +
    trailing_space_count +
    all_caps_count +
    lowercase_count +
    mixed_case_count
)

assert len(modified_rows) == expected_modified_rows, \
    "Tracker update failed."

print("\n✅ modified_rows tracker updated successfully.")
```

```python
#CELL 20 - HIDDEN TABS IMPLEMENTATION

# Number of rows to modify
hidden_tab_count = 25

# Select rows that have not been modified yet
available_rows = list(set(orders_dirty.index) - modified_rows)

selected_rows_hidden_tab = random.sample(
    available_rows,
    hidden_tab_count
)

# Store original values for validation
hidden_tab_before = orders_dirty.loc[
    selected_rows_hidden_tab,
    "Order_Status"
].copy()

# Split rows into leading and trailing tab groups
leading_tab_rows = selected_rows_hidden_tab[:13]
trailing_tab_rows = selected_rows_hidden_tab[13:]

# Inject leading tabs
orders_dirty.loc[
    leading_tab_rows,
    "Order_Status"
] = (
    "\t" +
    orders_dirty.loc[
        leading_tab_rows,
        "Order_Status"
    ]
)

# Inject trailing tabs
orders_dirty.loc[
    trailing_tab_rows,
    "Order_Status"
] = (
    orders_dirty.loc[
        trailing_tab_rows,
        "Order_Status"
    ] +
    "\t"
)
```

```python
#CELL 21 - HIDDEN TABS VALIDATION

print("=" * 60)
print("HIDDEN TAB VALIDATION")
print("=" * 60)

print(f"Expected Modified Rows : {hidden_tab_count}")
print(f"Actual Modified Rows   : {len(selected_rows_hidden_tab)}")

# Count rows containing tab characters
hidden_tab_validation = (
    orders_dirty.loc[
        selected_rows_hidden_tab,
        "Order_Status"
    ].str.contains("\t", regex=False).sum()
)

print(f"Rows Containing Hidden Tabs : {hidden_tab_validation}")

print("\nSample Validation (First 5 Rows)")
print("-" * 60)

for row in selected_rows_hidden_tab[:5]:

    print(f"\nRow Index : {row}")
    print(f"Before : {repr(hidden_tab_before.loc[row])}")
    print(f"After  : {repr(orders_dirty.loc[row, 'Order_Status'])}")

    print("\nComplete Row After Modification")
    print(orders_dirty.loc[[row]])

assert len(selected_rows_hidden_tab) == hidden_tab_count, \
    "Incorrect number of rows selected."

assert hidden_tab_validation == hidden_tab_count, \
    "Hidden tab validation failed."

print("\n✅ Hidden tabs successfully injected.")
```

```python
#CELL 22 - HIDDEN TABS TRACKER UPDATE

# Update modified rows tracker
modified_rows.update(selected_rows_hidden_tab)

print("=" * 60)
print("MODIFIED ROWS TRACKER UPDATED")
print("=" * 60)

print(f"Hidden Tabs Added To : {len(selected_rows_hidden_tab)} rows")
print(f"Total Modified Rows  : {len(modified_rows)}")

print("\nFirst 10 Modified Row Indexes:")
print(sorted(list(modified_rows))[:10])

# Validation
expected_modified_rows = (
    leading_space_count +
    trailing_space_count +
    all_caps_count +
    lowercase_count +
    mixed_case_count +
    hidden_tab_count
)

assert len(modified_rows) == expected_modified_rows, \
    "Tracker update failed."

print("\n✅ modified_rows tracker updated successfully.")
```

```python
#CELL 23 - BLANK VALUES IMPLEMENTATION

# Number of rows to modify
blank_value_count = 20

# Select eligible rows that have not been modified yet
available_rows = orders_dirty.loc[
    ~orders_dirty.index.isin(modified_rows)
]

# Select rows based on original status distribution
delivered_rows = available_rows[
    available_rows["Order_Status"] == "Delivered"
].sample(n=17, random_state=42).index.tolist()

cancelled_rows = available_rows[
    available_rows["Order_Status"] == "Cancelled"
].sample(n=2, random_state=42).index.tolist()

returned_rows = available_rows[
    available_rows["Order_Status"] == "Returned"
].sample(n=1, random_state=42).index.tolist()

selected_rows_blank = (
    delivered_rows +
    cancelled_rows +
    returned_rows
)

# Store original values
blank_before = orders_dirty.loc[
    selected_rows_blank,
    "Order_Status"
].copy()

# Inject blank values
orders_dirty.loc[
    selected_rows_blank,
    "Order_Status"
] = np.nan
```

```python
#CELL 24 - BLANK VALUES VALIDATION

print("=" * 60)
print("BLANK VALUES VALIDATION")
print("=" * 60)

print(f"Expected Modified Rows : {blank_value_count}")
print(f"Actual Modified Rows   : {len(selected_rows_blank)}")

# Count blank values
blank_validation = (
    orders_dirty.loc[
        selected_rows_blank,
        "Order_Status"
    ].isna().sum()
)

print(f"Blank Values Created : {blank_validation}")

print("\nSample Validation (First 5 Rows)")
print("-" * 60)

for row in selected_rows_blank[:5]:

    print(f"\nRow Index : {row}")
    print(f"Before : {repr(blank_before.loc[row])}")
    print(f"After  : {repr(orders_dirty.loc[row, 'Order_Status'])}")

    print("\nComplete Row After Modification")
    print(orders_dirty.loc[[row]])

assert len(selected_rows_blank) == blank_value_count, \
    "Incorrect number of rows selected."

assert blank_validation == blank_value_count, \
    "Blank value validation failed."

print("\n✅ Blank values successfully injected.")
```

```python
#CELL 25 - BLANK VALUES TRACKER UPDATE

# Update modified rows tracker
modified_rows.update(selected_rows_blank)

print("=" * 60)
print("MODIFIED ROWS TRACKER UPDATED")
print("=" * 60)

print(f"Blank Values Added To : {len(selected_rows_blank)} rows")
print(f"Total Modified Rows   : {len(modified_rows)}")

print("\nFirst 10 Modified Row Indexes:")
print(sorted(list(modified_rows))[:10])

# Validation
expected_modified_rows = (
    leading_space_count +
    trailing_space_count +
    all_caps_count +
    lowercase_count +
    mixed_case_count +
    hidden_tab_count +
    blank_value_count
)

assert len(modified_rows) == expected_modified_rows, \
    "Tracker update failed."

print("\n✅ modified_rows tracker updated successfully.")
```

```python
#CELL 26 - DUPLICATE RECORDS IMPLEMENTATION

# Number of duplicate records to create
duplicate_count = 25

# Calculate the most recent 25% of orders
recent_order_count = int(len(orders_dirty) * 0.25)

recent_orders = (
    orders_dirty
    .sort_values("Order_Date", ascending=False)
    .head(recent_order_count)
)

# Select duplicate rows deterministically
duplicate_rows = recent_orders.sample(
    n=duplicate_count,
    random_state=42
).copy()

# Store duplicated Order_IDs for validation
duplicated_order_ids = duplicate_rows["Order_ID"].tolist()

# Append duplicate rows
orders_dirty = pd.concat(
    [orders_dirty, duplicate_rows],
    ignore_index=True
)
```

```python
#CELL 27 - DUPLICATE RECORDS VALIDATION

print("=" * 60)
print("DUPLICATE RECORDS VALIDATION")
print("=" * 60)

print(f"Original Rows : {original_row_count}")
print(f"Duplicate Rows Added : {duplicate_count}")
print(f"Current Rows : {len(orders_dirty)}")

# Count duplicate Order_ID records
duplicate_validation = (
    orders_dirty["Order_ID"]
    .duplicated(keep=False)
    .sum()
)

print(f"\nDuplicate Order_ID Records : {duplicate_validation}")

print("\nSample Duplicate Order_IDs")
print("-" * 60)

for order_id in duplicated_order_ids[:5]:

    print(f"\nOrder_ID : {order_id}")

    print(
        orders_dirty[
            orders_dirty["Order_ID"] == order_id
        ]
    )

assert len(orders_dirty) == original_row_count + duplicate_count, \
    "Duplicate row generation failed."

assert duplicate_validation == duplicate_count * 2, \
    "Duplicate validation failed."

print("\n✅ Duplicate records successfully created.")
```

```python
#CELL 28 - SHUFFLE DATASET FINAL DUPLICATE DISTRIBUTION

# Shuffle dataset to distribute duplicate records
orders_dirty = (
    orders_dirty
    .sample(
        frac=1,
        random_state=42
    )
    .reset_index(drop=True)
)

print("=" * 60)
print("DATASET SHUFFLED")
print("=" * 60)

print(f"Final Row Count : {len(orders_dirty)}")

# Validation
assert len(orders_dirty) == original_row_count + duplicate_count, \
    "Dataset shuffle affected row count."

print("✅ Duplicate records distributed successfully.")
print("✅ Dataset shuffled.")
```

```python
#CELL 29 - FINAL DATASET INTEGRITY VALIDATION

print("=" * 60)
print("FINAL DATASET INTEGRITY VALIDATION")
print("=" * 60)

# Basic validation
current_shape = orders_dirty.shape
current_columns = orders_dirty.columns.tolist()

print(f"Master Dataset Shape : {original_shape}")
print(f"Dirty Dataset Shape  : {current_shape}")

print("\nChecking Column Structure...")
assert current_columns == original_columns, \
    "Column structure has changed."

print("✅ Column structure preserved.")

# Validate row count
expected_rows = original_row_count + duplicate_count

assert len(orders_dirty) == expected_rows, \
    "Final row count mismatch."

print("✅ Final row count validated.")

# Validate duplicate Order_IDs
duplicate_order_ids = (
    orders_dirty["Order_ID"]
    .duplicated(keep=False)
    .sum()
)

print(f"\nRows In Duplicate Pairs : {duplicate_order_ids}")

assert duplicate_order_ids == duplicate_count * 2, \
    "Unexpected duplicate Order_ID count."

print("✅ Duplicate records validated.")

# Validate master dataset unchanged
assert orders_master.shape == original_shape, \
    "Master dataset was modified."

assert orders_master.columns.tolist() == original_columns, \
    "Master dataset columns changed."

print("✅ Master dataset remains unchanged.")

print("\n🎉 Final integrity validation passed.")
```

```python
#CELL 30 - EXPORT Orders_Dirty.csv

output_file = "Orders_Dirty.csv"

orders_dirty.to_csv(
    output_file,
    index=False
)

print("=" * 60)
print("EXPORT COMPLETED")
print("=" * 60)

print(f"File Name : {output_file}")
print(f"Rows      : {len(orders_dirty)}")
print(f"Columns   : {orders_dirty.shape[1]}")

print("\n✅ Orders_Dirty.csv exported successfully.")
```
