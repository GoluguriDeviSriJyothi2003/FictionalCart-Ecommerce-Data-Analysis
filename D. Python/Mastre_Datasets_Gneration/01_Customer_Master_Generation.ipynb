```python
#CELL 2 - IMPORT LIBRARIES
import random
import numpy as np
import pandas as pd
```

```python
#CELL 3 - PROJECT CONFIGURATION
RANDOM_SEED = 42
TOTAL_CUSTOMERS = 750
random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
```

```python
#CELL 4 - CUSTOMER MASTER DATA

# -------------------------
# Customer Segments
# -------------------------

CUSTOMER_SEGMENTS = {
    "Regular": 0.50,
    "Budget": 0.35,
    "Premium": 0.15
}


# -------------------------
# Gender Distribution
# -------------------------

GENDER_DISTRIBUTION = {
    "Male": 0.49,
    "Female": 0.49,
    "Other": 0.02
}


# -------------------------
# City - State Mapping
# -------------------------

CITY_STATE = {
    "Hyderabad": "Telangana",
    "Visakhapatnam": "Andhra Pradesh",
    "Bengaluru": "Karnataka",
    "Mysuru": "Karnataka",
    "Chennai": "Tamil Nadu",
    "Coimbatore": "Tamil Nadu",
    "Kochi": "Kerala",
    "Thiruvananthapuram": "Kerala",
    "Mumbai": "Maharashtra",
    "Pune": "Maharashtra",
    "Nagpur": "Maharashtra",
    "Ahmedabad": "Gujarat",
    "Surat": "Gujarat",
    "Jaipur": "Rajasthan",
    "Udaipur": "Rajasthan",
    "New Delhi": "Delhi",
    "Lucknow": "Uttar Pradesh",
    "Kanpur": "Uttar Pradesh",
    "Kolkata": "West Bengal",
    "Bhubaneswar": "Odisha",
    "Patna": "Bihar",
    "Ranchi": "Jharkhand",
    "Guwahati": "Assam",
    "Chandigarh": "Chandigarh"
}
```

```python
#CELL 5 - CUSTOMER GEOGRAPHY

CITY_TIERS = {
    "Tier 1": [
        "Mumbai",
        "Bengaluru",
        "Hyderabad",
        "Chennai",
        "New Delhi",
        "Kolkata"
    ],

    "Tier 2": [
        "Pune",
        "Ahmedabad",
        "Visakhapatnam",
        "Kochi",
        "Lucknow",
        "Surat"
    ],

    "Tier 3": [
        "Mysuru",
        "Coimbatore",
        "Thiruvananthapuram",
        "Nagpur",
        "Jaipur",
        "Udaipur",
        "Kanpur",
        "Bhubaneswar",
        "Patna",
        "Ranchi",
        "Guwahati",
        "Chandigarh"
    ]
}


TIER_DISTRIBUTION = {
    "Tier 1": 0.50,
    "Tier 2": 0.30,
    "Tier 3": 0.20
}
```

```python
#CELL 6 - HELPER FUNCTIONS

def generate_customer_id(customer_number):
    """
    Generate a unique customer ID.

    Example:
    1 -> C0001
    25 -> C0025
    750 -> C0750
    """
    return f"C{customer_number:04d}"


def generate_gender():
    """
    Generate gender based on the
    predefined business distribution.
    """
    return np.random.choice(
        list(GENDER_DISTRIBUTION.keys()),
        p=list(GENDER_DISTRIBUTION.values())
    )


def generate_customer_segment():
    """
    Generate customer segment based on
    predefined business distribution.
    """
    return np.random.choice(
        list(CUSTOMER_SEGMENTS.keys()),
        p=list(CUSTOMER_SEGMENTS.values())
    )
```

```python
#CELL 7 - HELPER FUNCTION VALIDATION

# -------------------------
# Customer ID
# -------------------------

assert generate_customer_id(1) == "C0001"
assert generate_customer_id(25) == "C0025"
assert generate_customer_id(750) == "C0750"

# -------------------------
# Gender
# -------------------------

for _ in range(100):
    gender = generate_gender()
    assert gender in GENDER_DISTRIBUTION

# -------------------------
# Customer Segment
# -------------------------

for _ in range(100):
    segment = generate_customer_segment()
    assert segment in CUSTOMER_SEGMENTS

print("✅ Helper function validation passed.")
```

```python
#CELL 8 - GENERATE CUSTOMER CITY AND STATE

def generate_city_state():
    """
    Generate a valid City-State pair based on the
    predefined city tier distribution.
    """

    # Select a city tier
    selected_tier = np.random.choice(
        list(TIER_DISTRIBUTION.keys()),
        p=list(TIER_DISTRIBUTION.values())
    )

    # Select a city within the chosen tier
    city = random.choice(CITY_TIERS[selected_tier])

    # Get the corresponding state
    state = CITY_STATE[city]

    return city, state
```

```python
#CELL 9 - VALIDATE CITY AND STATE GENERATION

for _ in range(1000):

    city, state = generate_city_state()

    # City must exist in the master list
    assert city in CITY_STATE

    # State must match the city
    assert CITY_STATE[city] == state

print("✅ City-State validation passed.")
```

```python
#CELL 10 - GENERATE CUSTOMER AGE

def generate_age():
    """
    Generate a realistic customer age using a normal distribution.

    Business Rules:
    - Minimum Age : 18
    - Maximum Age : 70
    - Most customers fall between 25 and 45
    """

    while True:

        age = int(round(np.random.normal(loc=35, scale=10)))

        if 18 <= age <= 70:
            return age
```

```python
#CELL 11 - VALIDATE CUSTOMER AGE

ages = [generate_age() for _ in range(10000)]

# Business Rule Validation
assert min(ages) >= 18
assert max(ages) <= 70

print("Minimum Age :", min(ages))
print("Maximum Age :", max(ages))
print("Average Age :", round(np.mean(ages), 2))

print("\n✅ Age validation passed.")
```

```python
#CELL 12 - CUSTOMER NAME MASTER

MALE_FIRST_NAMES = [
    "Aarav", "Vivaan", "Aditya", "Vihaan", "Arjun",
    "Sai", "Krishna", "Rahul", "Kiran", "Rohan",
    "Ishaan", "Aryan", "Sanjay", "Akash", "Nikhil"
]

FEMALE_FIRST_NAMES = [
    "Ananya", "Diya", "Aadhya", "Anika", "Priya",
    "Sneha", "Pooja", "Kavya", "Meera", "Nisha",
    "Neha", "Shreya", "Ritika", "Deepika", "Swathi"
]

LAST_NAMES = [
    "Sharma", "Verma", "Reddy", "Patel", "Singh",
    "Gupta", "Nair", "Menon", "Iyer", "Rao",
    "Yadav", "Jain", "Kulkarni", "Joshi", "Das",
    "Chatterjee", "Mishra", "Sinha", "Pillai", "Bose",
    "Kapoor", "Malhotra", "Agarwal", "Naidu", "Gowda",
    "Shetty", "Kumar", "Bhat", "Pandey", "Roy"
]
```

```python
#CELL 13 - GENERATE CUSTOMER NAME

def generate_customer_name(gender):
    """
    Generate a realistic customer name
    based on the customer's gender.
    """

    if gender == "Male":
        first_name = random.choice(MALE_FIRST_NAMES)

    elif gender == "Female":
        first_name = random.choice(FEMALE_FIRST_NAMES)

    else:
        # Use either list for 'Other'
        first_name = random.choice(
            MALE_FIRST_NAMES + FEMALE_FIRST_NAMES
        )

    last_name = random.choice(LAST_NAMES)

    return f"{first_name} {last_name}"
```

```python
#CELL 14 - VALIDATE CUSTOMER NAME GENERATION

for gender in ["Male", "Female", "Other"]:

    for _ in range(100):

        customer_name = generate_customer_name(gender)

        assert isinstance(customer_name, str)
        assert len(customer_name.split()) == 2

print("✅ Customer name validation passed.")
```

```python
#CELL 15 - GENERATE CUSTOMER MASTER

customer_records = []

for customer_number in range(1, TOTAL_CUSTOMERS + 1):

    customer_id = generate_customer_id(customer_number)

    gender = generate_gender()
    
    customer_name = generate_customer_name(gender)

    age = generate_age()

    city, state = generate_city_state()

    customer_segment = generate_customer_segment()

    customer_records.append({
        "Customer_ID": customer_id,
        "Customer_Name": customer_name,
        "Gender": gender,
        "Age": age,
        "City": city,
        "State": state,
        "Customer_Segment": customer_segment
    })

customer_master = pd.DataFrame(customer_records)

print(f"Total Customers Generated : {len(customer_master)}")
customer_master.head()
```

```python
#CELL 16 - CUSTOMER MASTER VALIDATION

print("=" * 60)
print("CUSTOMER MASTER VALIDATION")
print("=" * 60)

# ------------------------------------------------------------
# Row Count
# ------------------------------------------------------------
print("\nRow Count Validation")
print("-" * 60)

assert len(customer_master) == TOTAL_CUSTOMERS

print(f"Total Customers : {len(customer_master)} ✅")


# ------------------------------------------------------------
# Customer ID Validation
# ------------------------------------------------------------
print("\nCustomer ID Validation")
print("-" * 60)

assert customer_master["Customer_ID"].is_unique

print("Customer IDs are unique ✅")


# ------------------------------------------------------------
# Missing Values
# ------------------------------------------------------------
print("\nMissing Values")
print("-" * 60)

print(customer_master.isnull().sum())


# ------------------------------------------------------------
# Gender Distribution
# ------------------------------------------------------------
print("\nGender Distribution")
print("-" * 60)

print(customer_master["Gender"].value_counts())


# ------------------------------------------------------------
# Customer Segment Distribution
# ------------------------------------------------------------
print("\nCustomer Segment Distribution")
print("-" * 60)

print(customer_master["Customer_Segment"].value_counts())


# ------------------------------------------------------------
# Age Statistics
# ------------------------------------------------------------
print("\nAge Statistics")
print("-" * 60)

print(customer_master["Age"].describe())


# ------------------------------------------------------------
# City Distribution (Top 10)
# ------------------------------------------------------------
print("\nTop 10 Cities")
print("-" * 60)

print(customer_master["City"].value_counts().head(10))


# ------------------------------------------------------------
# Validation Complete
# ------------------------------------------------------------
print("\n✅ Customer Master validation completed successfully.")
```

```python
#CELL 17 - RANDOM SAMPLE

customer_master.sample(10, random_state=RANDOM_SEED)
```

```python
#CELL 18 - GENERATE CUSTOMER JOIN DATE

from datetime import datetime, timedelta

PROJECT_START_DATE = datetime(2023, 1, 1)
PROJECT_END_DATE = datetime(2025, 12, 31)

def generate_join_date():
    """
    Generate a random customer join date
    between the project start and end dates.
    """

    total_days = (PROJECT_END_DATE - PROJECT_START_DATE).days

    random_days = random.randint(0, total_days)

    join_date = PROJECT_START_DATE + timedelta(days=random_days)

    return join_date.strftime("%Y-%m-%d")
```

```python
#CELL 19 - ADD JOIN DATE TO CUSTOMER MASTER

customer_master["Join_Date"] = [
    generate_join_date()
    for _ in range(len(customer_master))
]

print("✅ Join_Date added successfully.")

customer_master.head()
```

```python
#CELL 20 - VALIDATE JOIN DATE

join_dates = pd.to_datetime(customer_master["Join_Date"])

assert join_dates.min() >= pd.Timestamp("2023-01-01")
assert join_dates.max() <= pd.Timestamp("2025-12-31")

print("Earliest Join Date :", join_dates.min().date())
print("Latest Join Date   :", join_dates.max().date())

print("\n✅ Join_Date validation passed.")
```

```python
#CELL 21 - EXPORT CUSTOMER MASTER

customer_master.to_csv(
    "Customer_Master.csv",
    index=False
)

print("✅ Customer_Master.csv exported successfully.")
```
