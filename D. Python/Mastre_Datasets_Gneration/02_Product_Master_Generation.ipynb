```python
#CELL 20 - SELLING PRICE CONFIGURATION

CATEGORY_PRICE_RANGE = {
    "Electronics": (500, 50000),
    "Fashion": (300, 5000),
    "Home & Kitchen": (200, 10000),
    "Beauty": (150, 3000),
    "Grocery": (50, 1500),
    "Sports": (300, 15000),
    "Books": (200, 1500),
    "Toys": (250, 5000)
}
```

```python
#CELL 21 - PRODUCT RATING CONFIGURATION

CATEGORY_RATING_RANGE = {
    "Electronics": (3.8, 5.0),
    "Fashion": (3.6, 5.0),
    "Home & Kitchen": (3.7, 5.0),
    "Beauty": (3.9, 5.0),
    "Grocery": (4.0, 5.0),
    "Sports": (3.8, 5.0),
    "Books": (4.0, 5.0),
    "Toys": (3.7, 5.0)
}
```

```python
#CELL 22 - LAUNCH DATE CONFIGURATION

PRODUCT_LAUNCH_START = datetime(2020, 1, 1)
PRODUCT_LAUNCH_END = datetime(2025, 12, 31)
```

```python
#CELL 23 - VALIDATE BUSINESS CONFIGURATION

print("=" * 60)
print("BUSINESS CONFIGURATION VALIDATION")
print("=" * 60)

print(f"Price Configuration Categories  : {len(CATEGORY_PRICE_RANGE)}")
print(f"Rating Configuration Categories : {len(CATEGORY_RATING_RANGE)}")

assert set(CATEGORY_PRICE_RANGE.keys()) == set(CATEGORY_DISTRIBUTION.keys()), \
    "Price configuration categories do not match."

assert set(CATEGORY_RATING_RANGE.keys()) == set(CATEGORY_DISTRIBUTION.keys()), \
    "Rating configuration categories do not match."

assert PRODUCT_LAUNCH_START < PRODUCT_LAUNCH_END, \
    "Launch date configuration is invalid."

print(f"Launch Date Start : {PRODUCT_LAUNCH_START.date()}")
print(f"Launch Date End   : {PRODUCT_LAUNCH_END.date()}")

print("\n✅ Business configuration validation passed.")
```

```python
#CELL 24 - HELPER FUCNTION: GENERATE PRODUCT ID

def generate_product_id(number):
    """
    Generate a Product ID in the format P0001.
    """

    return f"P{number:04d}"
```

```python
#CELL 25 - VALIDATE PRODUCT ID FUNCTION

print("=" * 60)
print("PRODUCT ID FUNCTION VALIDATION")
print("=" * 60)

sample_ids = [
    generate_product_id(1),
    generate_product_id(25),
    generate_product_id(200)
]

for product_id in sample_ids:
    print(product_id)

assert generate_product_id(1) == "P0001"
assert generate_product_id(25) == "P0025"
assert generate_product_id(200) == "P0200"

print("\n✅ Product ID helper function validation passed.")
```

```python
#ECLL 26 - HELPER FUNCTION: GENERATE SELLING PRICE

def generate_price(category):
    """
    Generate a realistic selling price based on product category.
    Prices are rounded to the nearest ₹10.
    """

    min_price, max_price = CATEGORY_PRICE_RANGE[category]

    price = random.randint(min_price, max_price)

    return round(price, -1)
```

```python
#CELL 27 - HELPER FUNCTION: GENERATE PRODUCT RATING

def generate_rating(category):
    """
    Generate a customer rating based on product category.
    """

    min_rating, max_rating = CATEGORY_RATING_RANGE[category]

    rating = random.uniform(min_rating, max_rating)

    return round(rating, 1)
```

```python
#CELL 28 - HELPER FUNCTION: GENERATE LAUNCH DATE

def generate_launch_date():
    """
    Generate a random product launch date.
    """

    days = (PRODUCT_LAUNCH_END - PRODUCT_LAUNCH_START).days

    random_days = random.randint(0, days)

    return PRODUCT_LAUNCH_START + timedelta(days=random_days)
```

```python
#CELL 29 - VALIDATE ALL HELPER FUNCTIONS

print("=" * 60)
print("HELPER FUNCTION VALIDATION")
print("=" * 60)

print("\nSample Product IDs")
print("------------------------------")

for i in [1, 25, 200]:
    print(generate_product_id(i))

print("\nSample Prices")
print("------------------------------")

for category in CATEGORY_DISTRIBUTION.keys():
    print(f"{category:<20}: ₹{generate_price(category)}")

print("\nSample Ratings")
print("------------------------------")

for category in CATEGORY_DISTRIBUTION.keys():
    print(f"{category:<20}: {generate_rating(category)}")

print("\nSample Launch Dates")
print("------------------------------")

for _ in range(5):
    print(generate_launch_date().date())

print("\nRunning Assertions...")

assert generate_product_id(1) == "P0001"
assert generate_product_id(200) == "P0200"

for category in CATEGORY_DISTRIBUTION.keys():

    price = generate_price(category)
    rating = generate_rating(category)

    min_price, max_price = CATEGORY_PRICE_RANGE[category]
    min_rating, max_rating = CATEGORY_RATING_RANGE[category]

    assert min_price <= price <= max_price
    assert min_rating <= rating <= max_rating

print("\n✅ All helper functions validated successfully.")
```

```python
#CELL 30 - GENERATE PRODUCT MASTER DATASET

product_data = []

product_number = 1

for category, products in PRODUCT_CATALOG.items():

    for product_name in products:

        product_data.append({

            "Product_ID": generate_product_id(product_number),

            "Product_Name": product_name,

            "Category": category,

            "Selling_Price": generate_price(category),

            "Product_Rating": generate_rating(category),

            "Launch_Date": generate_launch_date().date()

        })

        product_number += 1

product_df = pd.DataFrame(product_data)
```

```python
#STEP 36 - CATEGORY DISTRIBUTION VALIDATION

print("=" * 60)
print("CATEGORY DISTRIBUTION VALIDATION")
print("=" * 60)

distribution = product_df["Category"].value_counts()

print(distribution)

for category, expected_count in CATEGORY_DISTRIBUTION.items():

    actual_count = distribution[category]

    assert actual_count == expected_count, \
        f"{category} count mismatch."

print("\n✅ Category distribution validation passed.")
```

```python
#CELL 37 - SELLING PRICE VALIDATION

print("=" * 60)
print("SELLING PRICE VALIDATION")
print("=" * 60)

for category, (min_price, max_price) in CATEGORY_PRICE_RANGE.items():

    category_prices = product_df.loc[
        product_df["Category"] == category,
        "Selling_Price"
    ]

    actual_min = category_prices.min()
    actual_max = category_prices.max()

    print(f"{category:<20} : ₹{actual_min:,}  →  ₹{actual_max:,}")

    assert actual_min >= min_price, \
        f"{category}: Minimum price is below the configured range."

    assert actual_max <= max_price, \
        f"{category}: Maximum price is above the configured range."

print("\n✅ Selling price validation passed.")
```

```python
#CELL 38 - PRODUCT RATING VALIDATION

print("=" * 60)
print("PRODUCT RATING VALIDATION")
print("=" * 60)

for category, (min_rating, max_rating) in CATEGORY_RATING_RANGE.items():

    category_ratings = product_df.loc[
        product_df["Category"] == category,
        "Product_Rating"
    ]

    actual_min = category_ratings.min()
    actual_max = category_ratings.max()

    print(f"{category:<20} : {actual_min:.1f}  →  {actual_max:.1f}")

    assert actual_min >= min_rating, \
        f"{category}: Minimum rating is below the configured range."

    assert actual_max <= max_rating, \
        f"{category}: Maximum rating is above the configured range."

print("\n✅ Product rating validation passed.")
```

```python
#CELL 39 - LAUNCH DATE VALIDATION

print("=" * 60)
print("LAUNCH DATE VALIDATION")
print("=" * 60)

earliest = min(product_df["Launch_Date"])
latest = max(product_df["Launch_Date"])

print(f"Earliest Launch Date : {earliest}")
print(f"Latest Launch Date   : {latest}")

assert earliest >= PRODUCT_LAUNCH_START.date(), \
    "Launch dates start before the configured range."

assert latest <= PRODUCT_LAUNCH_END.date(), \
    "Launch dates extend beyond the configured range."

print("\n✅ Launch date validation passed.")
```

```python
#CELL 40 - PRODUCT NAME VALIDATION

print("=" * 60)
print("PRODUCT NAME VALIDATION")
print("=" * 60)

duplicates = product_df["Product_Name"].duplicated().sum()

print(f"Duplicate Product Names : {duplicates}")

assert duplicates == 0, \
    "Duplicate product names found."

print("\n✅ Product name validation passed.")
```

```python
#CELL 41 - FINAL VALIDATION SUMMARY

print("=" * 60)
print("PRODUCT MASTER VALIDATION SUMMARY")
print("=" * 60)

print(f"Total Products : {len(product_df)}")
print(f"Total Columns  : {product_df.shape[1]}")

print("\nValidation Checklist")
print("-" * 60)

print("✅ Dataset Structure")
print("✅ Missing Values")
print("✅ Product ID Uniqueness")
print("✅ Category Distribution")
print("✅ Selling Price Range")
print("✅ Product Rating Range")
print("✅ Launch Date Range")
print("✅ Product Name Uniqueness")

print("\n🎉 Product Master validation completed successfully.")
```
